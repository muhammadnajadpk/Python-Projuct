```python
print("Total Applicants:", len(df))

print("Approved Applications:", (df["card"] == "yes").sum())

print("Rejected Applications:", (df["card"] == "no").sum())

print("Average Income:", df["income"].mean())

print("Average Age:", df["age"].mean())

print("Average Expenditure:", df["expenditure"].mean())
```
